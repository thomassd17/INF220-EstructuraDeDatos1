## 1) ADT Polinomio

Se representa el polinomio con un diccionario `{exponente: coeficiente}`, que es una forma simple de guardar solo los términos que existen (sin reservar espacio para exponentes intermedios en 0).

```python
class Polinomio:
    def __init__(self):
        self.terminos = {}   # exponente -> coeficiente

    def agregar_termino(self, coef, exp):
        if exp < 0:
            raise ValueError("El exponente no puede ser negativo")
        self.terminos[exp] = self.terminos.get(exp, 0) + coef
        if self.terminos[exp] == 0:
            del self.terminos[exp]   # se limpia el termino si el coeficiente queda en 0

    def evaluar(self, x):
        return sum(coef * (x ** exp) for exp, coef in self.terminos.items())

    def sumar(self, otro):
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            resultado.agregar_termino(coef, exp)
        for exp, coef in otro.terminos.items():
            resultado.agregar_termino(coef, exp)
        return resultado

    def restar(self, otro):
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            resultado.agregar_termino(coef, exp)
        for exp, coef in otro.terminos.items():
            resultado.agregar_termino(-coef, exp)
        return resultado

    def derivar(self):
        resultado = Polinomio()
        for exp, coef in self.terminos.items():
            if exp > 0:
                resultado.agregar_termino(coef * exp, exp - 1)
        return resultado

    def __str__(self):
        if not self.terminos:
            return "0"
        partes = []
        for exp in sorted(self.terminos.keys(), reverse=True):
            coef = self.terminos[exp]
            if exp == 0:
                partes.append(f"{coef}")
            elif exp == 1:
                partes.append(f"{coef}x")
            else:
                partes.append(f"{coef}x^{exp}")
        return " + ".join(partes).replace("+ -", "- ")


# --- Pruebas con al menos 3 polinomios distintos ---
p1 = Polinomio()
p1.agregar_termino(3, 2)   # 3x^2
p1.agregar_termino(2, 1)   # 2x
p1.agregar_termino(1, 0)   # 1

p2 = Polinomio()
p2.agregar_termino(5, 1)   # 5x
p2.agregar_termino(-4, 0)  # -4

p3 = Polinomio()
p3.agregar_termino(1, 3)   # x^3
p3.agregar_termino(-2, 2)  # -2x^2

print("p1 =", p1)
print("p2 =", p2)
print("p3 =", p3)

print("\np1 evaluado en x=2:", p1.evaluar(2))
print("p2 evaluado en x=2:", p2.evaluar(2))
print("p3 evaluado en x=2:", p3.evaluar(2))

print("\np1 + p2 =", p1.sumar(p2))
print("p1 - p2 =", p1.restar(p2))
print("derivada de p1 =", p1.derivar())
print("derivada de p3 =", p3.derivar())
```

## 2) ADT Conjunto

Se implementa `ConjuntoADT` usando una lista interna, validando la unicidad de los elementos manualmente en `agregar`.

```python
class ConjuntoADT:
    def __init__(self, elementos=None):
        self.elementos = []
        if elementos is not None:
            for elemento in elementos:
                self.agregar(elemento)

    def agregar(self, elemento):
        if elemento not in self.elementos:
            self.elementos.append(elemento)

    def quitar(self, elemento):
        if elemento in self.elementos:
            self.elementos.remove(elemento)

    def contiene(self, elemento):
        return elemento in self.elementos

    def union(self, otro):
        return ConjuntoADT(self.elementos + otro.elementos)

    def interseccion(self, otro):
        return ConjuntoADT([e for e in self.elementos if e in otro.elementos])

    def diferencia(self, otro):
        return ConjuntoADT([e for e in self.elementos if e not in otro.elementos])

    def diferencia_simetrica(self, otro):
        return self.diferencia(otro).union(otro.diferencia(self))

    def es_subconjunto(self, otro):
        return all(e in otro.elementos for e in self.elementos)

    def __len__(self):
        return len(self.elementos)

    def __str__(self):
        return "{" + ", ".join(map(str, self.elementos)) + "}"


# --- Caso de prueba 1: datos numericos ---
conjunto_a = ConjuntoADT([1, 2, 3, 4])
conjunto_b = ConjuntoADT([3, 4, 5, 6])

print("A =", conjunto_a)
print("B =", conjunto_b)
print("Union:", conjunto_a.union(conjunto_b))
print("Interseccion:", conjunto_a.interseccion(conjunto_b))
print("Diferencia (A-B):", conjunto_a.diferencia(conjunto_b))
print("Diferencia simetrica:", conjunto_a.diferencia_simetrica(conjunto_b))
print("Es A subconjunto de B?:", conjunto_a.es_subconjunto(conjunto_b))

# --- Caso de prueba 2: otro caso numerico (subconjunto real) ---
conjunto_c = ConjuntoADT([1, 2])
print("\nC =", conjunto_c)
print("Es C subconjunto de A?:", conjunto_c.es_subconjunto(conjunto_a))

# --- Caso de prueba 3: datos con cadenas ---
categorias_1 = ConjuntoADT(["Electronica", "Ropa", "Alimentos"])
categorias_2 = ConjuntoADT(["Ropa", "Hogar"])

print("\nCategorias 1 =", categorias_1)
print("Categorias 2 =", categorias_2)
print("Union:", categorias_1.union(categorias_2))
print("Interseccion:", categorias_1.interseccion(categorias_2))
print("Diferencia simetrica:", categorias_1.diferencia_simetrica(categorias_2))
```

## 3) Matriz dispersa

Se implementa `MatrizDispersa(filas, columnas)` usando un **diccionario de coordenadas** `{(i, j): valor}`, guardando solo las celdas distintas de cero.

```python
class MatrizDispersa:
    def __init__(self, filas, columnas):
        if filas <= 0 or columnas <= 0:
            raise ValueError("Filas y columnas deben ser mayores a 0")
        self.filas = filas
        self.columnas = columnas
        self.datos = {}   # (i, j) -> valor (solo se guardan los valores distintos de 0)

    def _validar_posicion(self, i, j):
        if not (0 <= i < self.filas) or not (0 <= j < self.columnas):
            raise IndexError("Posicion fuera de los limites de la matriz")

    def establecer(self, i, j, valor):
        self._validar_posicion(i, j)
        if valor == 0:
            self.datos.pop((i, j), None)   # no se guarda el cero, se elimina si existia
        else:
            self.datos[(i, j)] = valor

    def obtener(self, i, j):
        self._validar_posicion(i, j)
        return self.datos.get((i, j), 0)

    def transponer(self):
        traspuesta = MatrizDispersa(self.columnas, self.filas)
        for (i, j), valor in self.datos.items():
            traspuesta.establecer(j, i, valor)
        return traspuesta

    def densidad(self):
        total_celdas = self.filas * self.columnas
        return len(self.datos) / total_celdas

    def sumar(self, otra):
        """Desafio: suma de matrices dispersas, con validacion de dimensiones."""
        if self.filas != otra.filas or self.columnas != otra.columnas:
            raise ValueError("Las matrices deben tener las mismas dimensiones para sumarse")
        resultado = MatrizDispersa(self.filas, self.columnas)
        for (i, j), valor in self.datos.items():
            resultado.establecer(i, j, resultado.obtener(i, j) + valor)
        for (i, j), valor in otra.datos.items():
            resultado.establecer(i, j, resultado.obtener(i, j) + valor)
        return resultado

    def __str__(self):
        filas_texto = []
        for i in range(self.filas):
            fila = [str(self.obtener(i, j)) for j in range(self.columnas)]
            filas_texto.append(" ".join(fila))
        return "\n".join(filas_texto)


# --- Ejemplo de uso ---
matriz = MatrizDispersa(4, 4)
matriz.establecer(0, 0, 5)
matriz.establecer(1, 2, 3)
matriz.establecer(3, 3, 7)

print("Matriz dispersa:")
print(matriz)
print("\nValor en (1,2):", matriz.obtener(1, 2))
print("Densidad:", matriz.densidad())

print("\nTranspuesta:")
print(matriz.transponer())
```

### Eficiencia: matriz dispersa vs. matriz densa

Se compara el uso de memoria de guardar la misma información como **matriz densa** (lista de listas con todas las celdas, incluyendo ceros) contra la **matriz dispersa** (solo las celdas no-cero), para un caso con pocos valores distintos de cero.

```python
import sys

FILAS, COLUMNAS = 100, 100   # matriz grande, pero con muy pocos valores no-cero

# --- Version densa: se reserva memoria para las 100 x 100 celdas ---
matriz_densa = [[0] * COLUMNAS for _ in range(FILAS)]
matriz_densa[0][0] = 5
matriz_densa[50][99] = 3
matriz_densa[10][20] = 7

# --- Version dispersa: solo se guardan las 3 celdas no-cero ---
matriz_dispersa_grande = MatrizDispersa(FILAS, COLUMNAS)
matriz_dispersa_grande.establecer(0, 0, 5)
matriz_dispersa_grande.establecer(50, 99, 3)
matriz_dispersa_grande.establecer(10, 20, 7)

celdas_totales = FILAS * COLUMNAS
celdas_guardadas_dispersa = len(matriz_dispersa_grande.datos)

print(f"Celdas totales de la matriz: {celdas_totales}")
print(f"Celdas realmente guardadas en la version dispersa: {celdas_guardadas_dispersa}")
print(f"Densidad: {matriz_dispersa_grande.densidad():.5f} "
      f"({matriz_dispersa_grande.densidad()*100:.3f}% de celdas ocupadas)")

print(f"\nTamanio aproximado en memoria (matriz densa, lista de listas): "
      f"{sys.getsizeof(matriz_densa) + sum(sys.getsizeof(fila) for fila in matriz_densa)} bytes")
print(f"Tamanio aproximado en memoria (diccionario disperso): "
      f"{sys.getsizeof(matriz_dispersa_grande.datos)} bytes")

print("\nConclusion: con pocos valores distintos de cero, la matriz dispersa consume "
      "muchisima menos memoria, porque la matriz densa reserva espacio para las 10,000 "
      "celdas aunque casi todas sean 0.")
```

## 4) Integración — Inventario por sucursal y producto

Problema aplicado que combina los 3 ADT:
- **Conjunto** → categorías de productos disponibles.
- **Polinomio** → estimación simple de demanda futura (crecimiento en función del tiempo).
- **Matriz dispersa** → inventario por sucursal x producto (la mayoría de combinaciones no tienen stock, es decir, son cero).

```python
# --- Conjunto: categorias de productos que maneja la empresa ---
categorias = ConjuntoADT(["Electronica", "Ropa", "Alimentos", "Hogar"])
print("Categorias disponibles:", categorias)

# --- Polinomio: estimacion simple de demanda mensual de un producto ---
# demanda(t) = 2t^2 + 5t + 20 , donde t = mes (0 = mes actual)
demanda_estimada = Polinomio()
demanda_estimada.agregar_termino(2, 2)
demanda_estimada.agregar_termino(5, 1)
demanda_estimada.agregar_termino(20, 0)

print("\nFuncion de demanda estimada:", demanda_estimada)
for mes in range(4):
    print(f"  Demanda estimada para el mes {mes}: {demanda_estimada.evaluar(mes)} unidades")

# --- Matriz dispersa: inventario (filas = sucursales, columnas = productos) ---
# 5 sucursales, 6 productos. La mayoria de combinaciones estan en 0 (sin stock).
NUM_SUCURSALES = 5
NUM_PRODUCTOS = 6
inventario = MatrizDispersa(NUM_SUCURSALES, NUM_PRODUCTOS)

inventario.establecer(0, 1, 15)   # sucursal 0, producto 1: 15 unidades
inventario.establecer(2, 4, 8)    # sucursal 2, producto 4: 8 unidades
inventario.establecer(3, 1, 30)   # sucursal 3, producto 1: 30 unidades

print("\nInventario (filas=sucursales, columnas=productos):")
print(inventario)
print(f"\nDensidad del inventario: {inventario.densidad()*100:.1f}% de combinaciones con stock")

# --- Reporte combinado ---
print("\n--- Reporte ---")
print(f"La empresa maneja {len(categorias)} categorias: {categorias}")
print(f"Se espera vender {demanda_estimada.evaluar(3)} unidades del producto estimado en el mes 3")
print(f"Actualmente solo el {inventario.densidad()*100:.1f}% de las combinaciones sucursal-producto tienen stock")
```

## Desafío

- **Multiplicación de polinomios**: se agrega el método `multiplicar(otro)` a la clase `Polinomio`.
- **Suma de matrices dispersas con validación de dimensiones**: ya se implementó como el método `sumar(otra)` dentro de `MatrizDispersa` (ver Ejercicio 3), que lanza `ValueError` si las dimensiones no coinciden.

```python
def multiplicar_polinomios(self, otro):
    resultado = Polinomio()
    for exp1, coef1 in self.terminos.items():
        for exp2, coef2 in otro.terminos.items():
            resultado.agregar_termino(coef1 * coef2, exp1 + exp2)
    return resultado

Polinomio.multiplicar = multiplicar_polinomios   # se agrega el metodo a la clase ya definida


# --- Prueba: multiplicacion de polinomios ---
p1 = Polinomio()
p1.agregar_termino(1, 1)   # x
p1.agregar_termino(2, 0)   # 2

p2 = Polinomio()
p2.agregar_termino(1, 1)   # x
p2.agregar_termino(-3, 0)  # -3

print("p1 =", p1)
print("p2 =", p2)
print("p1 * p2 =", p1.multiplicar(p2))   # esperado: (x+2)(x-3) = x^2 - x - 6

assert p1.multiplicar(p2).evaluar(5) == p1.evaluar(5) * p2.evaluar(5)
print("Correcto: la evaluacion del producto coincide con el producto de las evaluaciones")


# --- Prueba: suma de matrices dispersas con validacion de dimensiones ---
matriz_x = MatrizDispersa(3, 3)
matriz_x.establecer(0, 0, 5)
matriz_x.establecer(1, 1, 2)

matriz_y = MatrizDispersa(3, 3)
matriz_y.establecer(0, 0, 3)
matriz_y.establecer(2, 2, 4)

print("\nMatriz X:")
print(matriz_x)
print("\nMatriz Y:")
print(matriz_y)

suma_matrices = matriz_x.sumar(matriz_y)
print("\nX + Y:")
print(suma_matrices)

matriz_z = MatrizDispersa(2, 2)   # dimensiones distintas, a proposito
try:
    matriz_x.sumar(matriz_z)
except ValueError as e:
    print("\nError controlado al sumar matrices de distinta dimension:", e)
```
