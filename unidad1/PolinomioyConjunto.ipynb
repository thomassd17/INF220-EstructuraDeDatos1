## Caso 1 — Conjunto dinámico (basado en clases)

**Cambios realizados respecto al original:**
- Se agregó el método `contiene(elemento)` y el soporte para el operador `in` mediante `__contains__`.
- Se agregó `__len__` para poder usar `len(conjunto)`.
- Se agregó `diferencia_simetrica()` (elementos que están en uno u otro conjunto, pero no en ambos).
- Se renombró el parámetro `otro_conjunto` a `otro` para simplificar la firma de los métodos.
- Se agregó una validación con `TypeError` si se intenta operar (`union`, `interseccion`, etc.) contra algo que no es un `Conjunto`.

```python
class Conjunto:
    def __init__(self, elementos=None):
        if elementos is None:
            self.elementos = []
        else:
            self.elementos = []
            for elemento in elementos:      # se recorre uno por uno para respetar unicidad desde el inicio
                self.agregar(elemento)

    def agregar(self, elemento):
        if elemento not in self.elementos:
            self.elementos.append(elemento)

    def quitar(self, elemento):
        if elemento in self.elementos:
            self.elementos.remove(elemento)

    def contiene(self, elemento):
        return elemento in self.elementos

    def __contains__(self, elemento):
        return self.contiene(elemento)

    def __len__(self):
        return len(self.elementos)

    def _validar(self, otro):
        if not isinstance(otro, Conjunto):
            raise TypeError("La operacion requiere otro objeto de tipo Conjunto")

    def union(self, otro):
        self._validar(otro)
        nuevo = Conjunto(self.elementos.copy())
        for elemento in otro.elementos:
            nuevo.agregar(elemento)
        return nuevo

    def interseccion(self, otro):
        self._validar(otro)
        nuevo = Conjunto()
        for elemento in self.elementos:
            if elemento in otro.elementos:
                nuevo.agregar(elemento)
        return nuevo

    def diferencia(self, otro):
        self._validar(otro)
        nuevo = Conjunto()
        for elemento in self.elementos:
            if elemento not in otro.elementos:
                nuevo.agregar(elemento)
        return nuevo

    def diferencia_simetrica(self, otro):
        self._validar(otro)
        return self.diferencia(otro).union(otro.diferencia(self))

    def __str__(self):
        return "{" + ", ".join(map(str, self.elementos)) + "}"


# --- Demostracion ---
conjunto1 = Conjunto([1, 2, 3])
conjunto2 = Conjunto([3, 4, 5])

print(f"Conjunto 1: {conjunto1}")
print(f"Conjunto 2: {conjunto2}")
print(f"Union: {conjunto1.union(conjunto2)}")
print(f"Interseccion: {conjunto1.interseccion(conjunto2)}")
print(f"Diferencia (1-2): {conjunto1.diferencia(conjunto2)}")
print(f"Diferencia simetrica: {conjunto1.diferencia_simetrica(conjunto2)}")

conjunto1.agregar(4)
print(f"Conjunto 1 tras agregar 4: {conjunto1}")

conjunto1.quitar(2)
print(f"Conjunto 1 tras quitar 2: {conjunto1}")

print(f"Tamanio de conjunto1: {len(conjunto1)}")
print(f"Contiene el 3?: {3 in conjunto1}")
```

## Caso 2 — Conjunto estático (basado en vectores/`set` de Python)

**Cambios realizados respecto al original:**
- El parámetro `elementos` ahora es opcional (por defecto `[]`), para poder crear un conjunto vacío sin error.
- Se agregó `__len__` para consultar el tamaño del conjunto.
- Se agregó `__eq__` para poder comparar dos conjuntos con `==` (ignorando el orden de los elementos).
- Se agregó `vaciar()` para limpiar todos los elementos.
- Se cambió el mensaje de `eliminar_elemento` para lanzar una advertencia con `warnings` en vez de solo un `print`, aunque se deja el comportamiento simple para no complicar el ADT.

```python
class ConjuntoEstatico:
    def __init__(self, elementos=None):
        if elementos is None:
            elementos = []
        if not isinstance(elementos, list):
            raise TypeError("Elementos debe ser una lista")
        self.elementos = elementos

    def agregar_elemento(self, elemento):
        if elemento not in self.elementos:
            self.elementos.append(elemento)

    def eliminar_elemento(self, elemento):
        if elemento in self.elementos:
            self.elementos.remove(elemento)
        else:
            print(f"Aviso: el elemento {elemento} no esta en el conjunto.")

    def vaciar(self):
        self.elementos = []

    def union(self, otro):
        return ConjuntoEstatico(list(set(self.elementos) | set(otro.elementos)))

    def interseccion(self, otro):
        return ConjuntoEstatico(list(set(self.elementos) & set(otro.elementos)))

    def diferencia(self, otro):
        return ConjuntoEstatico(list(set(self.elementos) - set(otro.elementos)))

    def es_subconjunto(self, otro):
        return set(self.elementos).issubset(set(otro.elementos))

    def contiene(self, elemento):
        return elemento in self.elementos

    def __len__(self):
        return len(self.elementos)

    def __eq__(self, otro):
        if not isinstance(otro, ConjuntoEstatico):
            return NotImplemented
        return set(self.elementos) == set(otro.elementos)

    def __str__(self):
        return str(self.elementos)


# --- Demostracion ---
vector_a = ConjuntoEstatico([5, 6, 7])
print("Conjunto estatico A:", vector_a)

vector_a.agregar_elemento(8)
print("A tras agregar 8:", vector_a)

vector_a.eliminar_elemento(6)
print("A tras eliminar 6:", vector_a)

vector_b = ConjuntoEstatico([7, 8, 9])

print("Union:", vector_a.union(vector_b))
print("Interseccion:", vector_a.interseccion(vector_b))
print("Diferencia:", vector_a.diferencia(vector_b))
print("Es subconjunto?:", vector_a.es_subconjunto(vector_b))
print("Contiene el 8?:", vector_a.contiene(8))
print("Tamanio de A:", len(vector_a))
print("A es igual a B?:", vector_a == vector_b)
print("A es igual a si mismo?:", vector_a == ConjuntoEstatico([8, 7, 5]))
```

## Caso 3 — Corrección y finalización del ADT incompleto

El código original tenía dos problemas:
1. Un **error de tipeo**: `def __ini__(self):` en vez de `def __init__(self):`, por lo que el constructor nunca se ejecutaba realmente al crear un objeto.
2. La clase no tenía **ningún método** para operar sobre `self.Elemento` (agregar, quitar, mostrar), solo declaraba los atributos.

**Cambios/cambios realizados:**
- Se corrigió `__ini__` a `__init__`.
- Se renombraron los atributos a minúsculas (`contexto`, `elementos`) siguiendo la convención de nombres de Python (PEP 8), manteniendo `nombre` igual.
- Se agregaron los métodos `agregar_elemento`, `eliminar_elemento` y `mostrar_info`, para que la clase sea funcional.
- Se conserva el ejemplo original de vector/lista con pequeñas variaciones de nombres.

```python
# --- Primera parte: demostracion vector -> lista (igual idea que el original, con variacion de nombres) ---
vector_base = [100]
vector_base.append(1)
vector_base.append(2)
vector_base.append(3)

lista_dinamica = list(vector_base)   # se crea una lista dinamica a partir del vector fijo
lista_dinamica.append(4)
lista_dinamica.append(5)
lista_dinamica.append(6)

print("Vector base (sin modificar):", vector_base)
print("Lista dinamica derivada:", lista_dinamica)


# --- Segunda parte: clase Conjunto con metadatos, corregida y completada ---
class ConjuntoConMetadatos:
    """Version corregida y completada del ADT 'conjunto' del caso 3."""

    def __init__(self, nombre=None, contexto=None):
        self.nombre = nombre
        self.contexto = contexto
        self.elementos = []   # antes 'self.Elemento', renombrado a minusculas

    def agregar_elemento(self, elemento):
        if elemento not in self.elementos:
            self.elementos.append(elemento)

    def eliminar_elemento(self, elemento):
        if elemento in self.elementos:
            self.elementos.remove(elemento)

    def mostrar_info(self):
        print(f"Nombre: {self.nombre}")
        print(f"Contexto: {self.contexto}")
        print(f"Elementos: {self.elementos}")


# --- Demostracion ---
mi_conjunto = ConjuntoConMetadatos(nombre="Numeros de prueba", contexto="Unidad 1 - Conjuntos")
mi_conjunto.agregar_elemento(100)
mi_conjunto.agregar_elemento(1)
mi_conjunto.agregar_elemento(2)
mi_conjunto.mostrar_info()

mi_conjunto.eliminar_elemento(1)
print("Despues de eliminar el 1:")
mi_conjunto.mostrar_info()
```

## Caso 4 — ADT Polinomio (lista enlazada)

**Cambios realizados respecto al original:**
- Se agregó `evaluar(x)` para calcular el valor numérico del polinomio dado un valor de `x`.
- Se agregó `grado()` para obtener el exponente más alto del polinomio.
- Se agregó `sumar(otro)` para sumar dos polinomios y devolver un `Polinomio` nuevo.
- Se agregó `multiplicar_escalar(k)` para multiplicar todos los coeficientes por una constante.
- Se ajustaron nombres de variables internas (`actual`/`anterior` se mantienen, pero se documentan con comentarios más claros).

```python
class NodoPolinomio:
    def __init__(self, coeficiente, exponente):
        self.coeficiente = coeficiente
        self.exponente = exponente
        self.siguiente = None


class Polinomio:
    def __init__(self):
        self.cabeza = None

    def agregar_termino(self, coeficiente, exponente):
        nuevo_nodo = NodoPolinomio(coeficiente, exponente)
        if self.cabeza is None:
            self.cabeza = nuevo_nodo
        else:
            actual = self.cabeza
            anterior = None
            while actual is not None and actual.exponente > exponente:
                anterior = actual
                actual = actual.siguiente
            if anterior is None:
                nuevo_nodo.siguiente = self.cabeza
                self.cabeza = nuevo_nodo
            else:
                anterior.siguiente = nuevo_nodo
                nuevo_nodo.siguiente = actual

    def grado(self):
        """Devuelve el mayor exponente del polinomio (0 si esta vacio)."""
        if self.cabeza is None:
            return 0
        return self.cabeza.exponente   # como esta ordenado de mayor a menor exponente

    def evaluar(self, x):
        """Evalua el polinomio para un valor numerico de x."""
        resultado = 0
        actual = self.cabeza
        while actual is not None:
            resultado += actual.coeficiente * (x ** actual.exponente)
            actual = actual.siguiente
        return resultado

    def sumar(self, otro):
        """Suma este polinomio con 'otro' y devuelve un Polinomio nuevo."""
        resultado = Polinomio()
        actual = self.cabeza
        while actual is not None:
            resultado.agregar_termino(actual.coeficiente, actual.exponente)
            actual = actual.siguiente

        actual_otro = otro.cabeza
        while actual_otro is not None:
            # Si ya existe un termino con ese exponente, se suman los coeficientes
            nodo_existente = resultado.cabeza
            encontrado = False
            while nodo_existente is not None:
                if nodo_existente.exponente == actual_otro.exponente:
                    nodo_existente.coeficiente += actual_otro.coeficiente
                    encontrado = True
                    break
                nodo_existente = nodo_existente.siguiente
            if not encontrado:
                resultado.agregar_termino(actual_otro.coeficiente, actual_otro.exponente)
            actual_otro = actual_otro.siguiente

        return resultado

    def multiplicar_escalar(self, k):
        """Devuelve un nuevo Polinomio con todos los coeficientes multiplicados por k."""
        resultado = Polinomio()
        actual = self.cabeza
        while actual is not None:
            resultado.agregar_termino(actual.coeficiente * k, actual.exponente)
            actual = actual.siguiente
        return resultado

    def __str__(self):
        if self.cabeza is None:
            return "0"

        resultado = ""
        actual = self.cabeza
        while actual is not None:
            if actual.coeficiente != 0:
                if actual.coeficiente > 0 and len(resultado) > 0:
                    resultado += "+"
                if actual.coeficiente == -1 and actual.exponente != 0:
                    resultado += "-"
                elif actual.coeficiente != 1 or actual.exponente == 0:
                    resultado += str(actual.coeficiente)
                if actual.exponente > 0:
                    resultado += "x"
                    if actual.exponente > 1:
                        resultado += "^" + str(actual.exponente)
            actual = actual.siguiente
        return resultado if resultado else "0"


# --- Demostracion ---
polinomio = Polinomio()
polinomio.agregar_termino(3, 2)
polinomio.agregar_termino(2, 1)
polinomio.agregar_termino(1, 0)
polinomio.agregar_termino(-1, 3)

print("Polinomio 1:", polinomio)
print("Grado:", polinomio.grado())
print("Evaluado en x=2:", polinomio.evaluar(2))

polinomio2 = Polinomio()
polinomio2.agregar_termino(0, 2)
polinomio2.agregar_termino(5, 1)
print("Polinomio 2:", polinomio2)

suma = polinomio.sumar(polinomio2)
print("Suma de polinomio1 + polinomio2:", suma)

escalado = polinomio2.multiplicar_escalar(3)
print("Polinomio 2 multiplicado por 3:", escalado)
```
